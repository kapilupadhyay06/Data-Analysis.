```python
from io import StringIO
import pandas as pd

data = StringIO('''
Roll_No,Branch,Gender,Study_Hours,Attendance,Maths,Python,ML
1,ECE,Female,3.5,70,77,50,88
2,CSE,Female,5.0,84,82,78,94
3,ECE,Female,3.6,88,76,80,77
4,ECE,Male,3.4,71,67,69,82
5,CSE,Male,5.9,64,86,80,75
6,CSE,Male,3.6,69,81,72,64
7,ECE,Male,0.5,84,66,75,74
8,IT,Female,4.5,76,83,81,60
9,ECE,Male,3.4,83,72,74,61
10,ECE,Male,3.6,96,93,73,93
''')

df = pd.read_csv(data)

df["Total"] = df[["Maths", "Python", "ML"]].sum(axis=1)

df["Percentage"] = (df["Total"] / 3).round(1)

df["Grade"] = "D"

df.loc[df["Percentage"] >= 60, "Grade"] = "C"

df.loc[df["Percentage"] >= 70, "Grade"] = "B"

df.loc[df["Percentage"] >= 80, "Grade"] = "A"

df.head()

print(df)
```

```python
counts = df["Branch"].value_counts()
plt.bar(counts.index, counts.values)

plt.xlabel("Branch")
plt.ylabel("Number of Students")
plt.title("Students per Branch")
plt.show()
```

```python
counts = df["Branch"].value_counts()

plt.bar(counts.index, counts.values, color=["red", "blue", "green", "orange"])

plt.xlabel("Branch")
plt.ylabel("Number of Students")
plt.title("Students per Branch")
plt.show()
```

```python
# draw a line plot of the pythoonn marks of the first 5 students (x= Roll_No).use a read dashed line with circle makers
first_5 = df.head(5)

plt.plot(first_5["Roll_No"], first_5["Python"], color="green", linestyle="--", marker="o")
plt.xlabel("Roll_No")
plt.ylabel("Python Marks")
plt.title("Python Marks of First 5 Students")
plt.show()
```

```python
first_5 = df.head(5)

plt.plot(first_5["Roll_No"], first_5["Python"], color="red", linestyle="--", marker="o")
plt.xticks([1, 2, 3, 4, 5], ["One", "Two", "Three", "Four", "Five"])
plt.xlabel("Roll_No")
plt.ylabel("Python Marks")
plt.title("Python Marks of First 5 Students")
plt.show()
```

```python
# in one graph, plot maths and python marks of the first 7 students as two differnt line add a legend
first_7 = df.head(7)

plt.plot(first_7["Roll_No"], first_7["Maths"], marker="o", label="Maths")
plt.plot(first_7["Roll_No"], first_7["Python"], marker="s", label="Python")
plt.xticks(first_7["Roll_No"])
plt.xlabel("Roll_No")
plt.ylabel("Marks")
plt.title("Maths vs Python Marks of First 7 Students")
plt.legend()
plt.show()
```

```python
# draw a histogram of attendence with 10 bins , orange colour and white edges
plt.hist(df["Attendance"], bins=10, color="orange", edgecolor="white")
plt.xlabel("Attendance")
plt.ylabel("Frequency")
plt.title("Attendance Distribution")
plt.show()
```
